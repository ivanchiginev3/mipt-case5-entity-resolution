```python
import pandas as pd
from pathlib import Path
```

```python
# ============================================================
# PATHS
# ============================================================

PROCESSED_DIR = Path("../data/processed")
OUTPUT_DIR = Path("../data/final")

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

DEDUP_PATH = PROCESSED_DIR / "deduplicated_sample.parquet"
CANONICAL_PATH = PROCESSED_DIR / "canonical_entities_full_sample.parquet"

print(OUTPUT_DIR)
```

```python
# ============================================================
# LOAD FINAL ARTIFACTS
# ============================================================

deduplicated = pd.read_parquet(DEDUP_PATH)
canonical_entities = pd.read_parquet(CANONICAL_PATH)

print(f"Deduplicated rows: {len(deduplicated):,}")
print(f"Canonical entities: {len(canonical_entities):,}")
```

```python
# ============================================================
# BUILD CUSTOMER-FACING DEDUPLICATED FILE
# ============================================================

deduplicated_export_cols = [
    "record_id",
    "country",
    "source_snapshot_date",
    "relation_kind",
    "relation_role",
    "party_type",
    "company_public_id",
    "company_name_norm",
    "party_public_id",
    "party_name",
    "name_latin",
    "entity_id",
    "cluster_size",
    "canonical_name",
]

deduplicated_export_cols = [
    col for col in deduplicated_export_cols
    if col in deduplicated.columns
]

deduplicated_export = deduplicated[
    deduplicated_export_cols
].copy()

display(deduplicated_export.head())

print(f"Rows: {len(deduplicated_export):,}")
print(f"Columns: {len(deduplicated_export.columns)}")
```

```python
# ============================================================
# BUILD CUSTOMER-FACING CANONICAL ENTITIES FILE
# ============================================================

canonical_export_cols = [
    "entity_id",
    "canonical_name",
    "cluster_size",
    "canonical_country",
    "canonical_script",
    "main_relation_kind",
    "main_relation_role",
    "unique_names_count",
    "unique_countries_count",
    "unique_companies_count",
    "is_singleton",
    "is_large_cluster",
    "is_suspicious_entity",
    "original_names",
    "normalized_names",
    "countries",
    "companies",
    "record_ids",
]

canonical_export_cols = [
    col for col in canonical_export_cols
    if col in canonical_entities.columns
]

canonical_export = canonical_entities[
    canonical_export_cols
].copy()

display(canonical_export.head())

print(f"Rows: {len(canonical_export):,}")
print(f"Columns: {len(canonical_export.columns)}")
```

```python
# ============================================================
# CLUSTER DISTRIBUTION REPORT
# ============================================================

cluster_distribution = (
    canonical_export
    .groupby("cluster_size")
    .size()
    .reset_index(name="entities_count")
    .sort_values("cluster_size")
)

display(cluster_distribution.head(30))

print("Largest clusters:")
display(
    canonical_export
    .sort_values("cluster_size", ascending=False)
    .head(20)
)
```

```python
# ============================================================
# FINAL SUMMARY
# ============================================================

summary = {
    "input_records": len(deduplicated_export),
    "unique_entities": canonical_export["entity_id"].nunique(),
    "singleton_entities": int((canonical_export["cluster_size"] == 1).sum()),
    "non_singleton_entities": int((canonical_export["cluster_size"] > 1).sum()),
    "records_in_duplicate_clusters": int(
        canonical_export.loc[
            canonical_export["cluster_size"] > 1,
            "cluster_size",
        ].sum()
    ),
    "largest_cluster_size": int(canonical_export["cluster_size"].max()),
    "deduplication_compression": float(
        1 - canonical_export["entity_id"].nunique() / len(deduplicated_export)
    ),
}

summary_df = pd.DataFrame([summary])

display(summary_df)
```

```python
# ============================================================
# SAVE FINAL FILES
# ============================================================

deduplicated_parquet_path = OUTPUT_DIR / "deduplicated_records.parquet"
deduplicated_csv_path = OUTPUT_DIR / "deduplicated_records.csv"

canonical_parquet_path = OUTPUT_DIR / "canonical_entities.parquet"
canonical_csv_path = OUTPUT_DIR / "canonical_entities.csv"

cluster_distribution_path = OUTPUT_DIR / "cluster_distribution.csv"
summary_path = OUTPUT_DIR / "final_summary.csv"

deduplicated_export.to_parquet(
    deduplicated_parquet_path,
    index=False,
)

deduplicated_export.to_csv(
    deduplicated_csv_path,
    index=False,
    encoding="utf-8-sig",
)

canonical_export.to_parquet(
    canonical_parquet_path,
    index=False,
)

canonical_export.to_csv(
    canonical_csv_path,
    index=False,
    encoding="utf-8-sig",
)

cluster_distribution.to_csv(
    cluster_distribution_path,
    index=False,
    encoding="utf-8-sig",
)

summary_df.to_csv(
    summary_path,
    index=False,
    encoding="utf-8-sig",
)

print("Saved final files:")
print(deduplicated_parquet_path)
print(deduplicated_csv_path)
print(canonical_parquet_path)
print(canonical_csv_path)
print(cluster_distribution_path)
print(summary_path)
```

```python
# ============================================================
# SANITY CHECK
# ============================================================

assert len(deduplicated_export) == summary["input_records"]
assert canonical_export["entity_id"].nunique() == summary["unique_entities"]
assert deduplicated_export["entity_id"].isna().sum() == 0

print("Sanity check passed.")
```

## Вывод по финальным файлам

На этом этапе были сформированы итоговые файлы для передачи заказчику.

Сформированы:

1. `deduplicated_records.parquet` / `deduplicated_records.csv`  
   Исходные записи с назначенным `entity_id`.

2. `canonical_entities.parquet` / `canonical_entities.csv`  
   Таблица уникальных сущностей.

3. `cluster_distribution.csv`  
   Распределение размеров кластеров.

4. `final_summary.csv`  
   Краткая статистика результата дедупликации.

Все записи с одинаковым `entity_id` считаются дубликатами одной сущности.
