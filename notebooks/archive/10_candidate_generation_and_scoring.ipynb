```python
import pandas as pd
import joblib
import sys

sys.path.append('..')
from src.pipeline.inference import run_inference_pipeline
```

```python
# ============================================================
# LOAD DATA AND MODEL
# ============================================================

df = pd.read_parquet("../data/processed/processed_dataset.parquet")

model = joblib.load(
    "../data/artifacts/random_forest_matcher.joblib"
)

print(f"Dataset rows: {len(df):,}")
print("Model loaded.")
```

```python
# ============================================================
# SAMPLE FOR FIRST FULL PIPELINE TEST
# ============================================================

# Для первого запуска берем sample.
# На всем датасете может быть слишком много кандидатов.

df_sample = df.sample(
    100_000,
    random_state=42,
)

print(f"Sample rows: {len(df_sample):,}")
```

```python
# ============================================================
# RUN INFERENCE PIPELINE
# ============================================================

scored_pairs = run_inference_pipeline(
    df=df_sample,
    model=model,
    max_block_size=500,
    score_threshold=0.5,
)

print(f"Scored candidate pairs: {len(scored_pairs):,}")

display(scored_pairs.head())
```

```python
pd.set_option('display.max_columns', None)
```

```python
scored_pairs.head()
```

```python
# ============================================================
# SCORE DISTRIBUTION
# ============================================================

print(scored_pairs["duplicate_score"].describe())

print("\nPredictions:")
print(scored_pairs["duplicate_prediction"].value_counts())
```

```python
# ============================================================
# TOP DUPLICATE CANDIDATES
# ============================================================

top_duplicates = (
    scored_pairs
    .sort_values("duplicate_score", ascending=False)
    .head(50)
)

display(
    top_duplicates[
        [
            "party_name_1",
            "party_name_2",
            "name_latin_1",
            "name_latin_2",
            "duplicate_score",
            "duplicate_prediction",
        ]
    ]
)
```

```python
# ============================================================
# PREDICTED DUPLICATES ONLY
# ============================================================

predicted_duplicates = scored_pairs[
    scored_pairs["duplicate_prediction"] == 1
].copy()

print(f"Predicted duplicate pairs: {len(predicted_duplicates):,}")

display(
    predicted_duplicates[
        [
            "party_name_1",
            "party_name_2",
            "duplicate_score",
        ]
    ]
    .sort_values("duplicate_score", ascending=False)
    .head(50)
)
```

```python
# ============================================================
# SAVE SCORED PAIRS
# ============================================================

save_path = "../data/processed/scored_pairs_sample.parquet"

scored_pairs.to_parquet(
    save_path,
    index=False,
)

print(f"Saved to: {save_path}")
```
