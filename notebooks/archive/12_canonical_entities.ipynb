```python
import pandas as pd
import sys

sys.path.append('..')
from src.models.canonicalization import (
    build_canonical_entities,
    add_entity_quality_flags,
    entity_summary,
)
```

```python
# ============================================================
# LOAD FILTERED CLUSTERED SAMPLE
# ============================================================

clustered_df = pd.read_parquet(
    "../data/processed/clustered_sample_filtered.parquet"
)

cluster_stats = pd.read_parquet(
    "../data/processed/cluster_stats_sample_filtered.parquet"
)

print(f"Clustered records: {len(clustered_df):,}")
print(f"Cluster stats rows: {len(cluster_stats):,}")

display(clustered_df.head())
```

```python
# ============================================================
# BUILD CANONICAL ENTITIES
# ============================================================

entities = build_canonical_entities(clustered_df)

entities = add_entity_quality_flags(
    entities,
    large_cluster_threshold=20,
    max_unique_names=5,
    max_unique_countries=3,
)

display(entities.head())
```

```python
# ============================================================
# ENTITY SUMMARY
# ============================================================

entity_summary(entities)
```

```python
# ============================================================
# SINGLETON VS NON-SINGLETON
# ============================================================

print("Singletons:")
print((entities["cluster_size"] == 1).sum())

print("\nNon-singleton entities:")
print((entities["cluster_size"] > 1).sum())

print("\nRecords covered by non-singleton entities:")
print(
    entities.loc[
        entities["cluster_size"] > 1,
        "cluster_size"
    ].sum()
)
```

```python
# ============================================================
# CHECK SUSPICIOUS ENTITIES
# ============================================================

suspicious_entities = entities[
    entities["is_suspicious_entity"] == 1
].copy()

print(f"Suspicious entities: {len(suspicious_entities):,}")

display(
    suspicious_entities[
        [
            "entity_id",
            "canonical_name",
            "cluster_size",
            "unique_names_count",
            "unique_countries_count",
            "original_names",
        ]
    ]
    .head(20)
)
```

```python
# ============================================================
# EXAMPLES OF CANONICAL ENTITIES
# ============================================================

examples = entities[
    entities["cluster_size"] > 1
].sample(
    10,
    random_state=42,
)

for _, row in examples.iterrows():
    print("=" * 80)
    print(f"entity_id: {row['entity_id']}")
    print(f"canonical_name: {row['canonical_name']}")
    print(f"cluster_size: {row['cluster_size']}")
    print(f"country: {row['canonical_country']}")

    print("\nOriginal names:")
    for name in row["original_names"][:10]:
        print("-", name)
```

```python
# ============================================================
# SAVE CANONICAL ENTITIES
# ============================================================

entities.to_parquet(
    "../data/processed/canonical_entities_sample.parquet",
    index=False,
)

print("Saved:")
print("../data/processed/canonical_entities_sample.parquet")
```
