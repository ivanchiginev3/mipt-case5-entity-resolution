```python
import pandas as pd
import sys

sys.path.append('..')
from src.features.augmentation import augment_train_pairs
```

```python
# ============================================================
# LOAD SPLITS
# ============================================================

train_pairs = pd.read_parquet("../data/processed/train_pairs.parquet")
valid_pairs = pd.read_parquet("../data/processed/valid_pairs.parquet")
test_pairs = pd.read_parquet("../data/processed/test_pairs.parquet")

print(f"Train: {len(train_pairs):,}")
print(f"Valid: {len(valid_pairs):,}")
print(f"Test:  {len(test_pairs):,}")
```

```python
# ============================================================
# BEFORE AUGMENTATION
# ============================================================

print("=" * 60)
print("TRAIN BEFORE AUGMENTATION")
print("=" * 60)

print(train_pairs["label"].value_counts())
print("\nPair types:")
print(train_pairs["pair_type"].value_counts())
```

```python
# ============================================================
# APPLY AUGMENTATION ONLY TO TRAIN
# ============================================================

train_augmented = augment_train_pairs(
    train_pairs=train_pairs,
    n_positive_aug=30_000,
    n_hard_negative_aug=30_000,
    random_state=42,
)

print("=" * 60)
print("TRAIN AFTER AUGMENTATION")
print("=" * 60)

print(f"Rows before: {len(train_pairs):,}")
print(f"Rows after:  {len(train_augmented):,}")

print("\nLabel distribution:")
print(train_augmented["label"].value_counts())

print("\nPair types:")
print(train_augmented["pair_type"].value_counts())
```

```python
# ============================================================
# AUGMENTED POSITIVE EXAMPLES
# ============================================================

display(
    train_augmented[
        train_augmented["pair_type"] == "augmented_positive"
    ][
        [
            "party_name_1",
            "party_name_2",
            "name_latin_1",
            "name_latin_2",
            "label",
            "pair_type",
        ]
    ]
    .head(30)
)
```

```python
# ============================================================
# AUGMENTED HARD NEGATIVE EXAMPLES
# ============================================================

display(
    train_augmented[
        train_augmented["pair_type"] == "augmented_hard_negative"
    ][
        [
            "party_name_1",
            "party_name_2",
            "name_latin_1",
            "name_latin_2",
            "label",
            "pair_type",
        ]
    ]
    .head(30)
)
```

```python
# ============================================================
# SAVE AUGMENTED TRAIN
# ============================================================

save_path = "../data/processed/train_pairs_augmented.parquet"

train_augmented.to_parquet(save_path, index=False)

print(f"Saved to: {save_path}")
print(f"Rows: {len(train_augmented):,}")
```

## Вывод по аугментации

Аугментация была применена только к обучающей выборке. Это позволяет не искажать проверочную и тестовую выборки.

После аугментации размер train увеличился со 120 894 до 180 894 пар. Были добавлены искусственные положительные и сложные отрицательные примеры.

При этом часть искусственных примеров может быть спорной: некоторые искажения слишком сильные, а некоторые отрицательные примеры могут теоретически оказаться вариантами одной сущности. Поэтому текущая аугментация рассматривается как первая версия и будет использоваться осторожно.
